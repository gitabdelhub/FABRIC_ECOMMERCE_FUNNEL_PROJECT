```python
from pyspark.sql.functions import current_timestamp, input_file_name

# 1. Lecture distribuée de tous les CSV bruts (Multi-mois / Scalable)
df_raw = (spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("Files/raw/*.csv")
)

# 2. Ajout des métadonnées d'audit technique
df_bronze = (df_raw
    .withColumn("_ingested_at", current_timestamp())
    .withColumn("_source_file", input_file_name())
)

# 3. Écriture Delta dans la couche Bronze avec écrasement sécurisé du schéma
(df_bronze.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("bronze_events")
)

print(f" Couche BRONZE terminée : {spark.table('bronze_events').count():,} lignes ingérées.")
```
